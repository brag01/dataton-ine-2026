# 01 · Validación de la API del INE Guatemala

Este notebook consulta el catálogo CKAN. **No presupone que todos los recursos del IPC admitan DataStore.**

```python
from ine_gt import INEClient
import pandas as pd
```

```python
ine = INEClient()
catalogo = ine.buscar_conjuntos("IPC")
print(f"Conjuntos encontrados: {len(catalogo)}")
display(catalogo.head(20))
```

```python
if catalogo.empty:
    print("No se encontraron conjuntos para IPC; probar otras palabras clave.")
else:
    recursos = ine.recursos(catalogo.iloc[0]["id"])
    display(recursos)
```

```python
if not catalogo.empty and not recursos.empty:
    activos = recursos[recursos["datastore_active"] == True]
    if activos.empty:
        print("Este conjunto no contiene recursos DataStore activos. Revisar formato y URL.")
    else:
        recurso_id = activos.iloc[0]["id"]
        muestra = ine.datos(recurso_id, max_rows=10)
        print("ID de recurso:", recurso_id)
        display(muestra)
        print("Columnas:", list(muestra.columns))
        print("Origen:", muestra.attrs.get("source"))
```

```python
ine.close()
```
